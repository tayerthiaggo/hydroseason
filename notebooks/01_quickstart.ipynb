# HydroSeason: Quickstart

**What this shows:** the one-call workflow — `run_hydroseason` on a monthly
water-extent CSV, producing an automatic route decision plus a
self-contained HTML report and four CSVs.

**Requirements:** core install only (`pip install hydroseason`) — no
`[raster]` or `[stac]` extras needed.

**Network:** none. Runs fully offline against a committed case-study CSV.

**Runtime:** a few seconds.

See also: [02_dea_acquisition.ipynb](02_dea_acquisition.ipynb) (fetch from DEA),
[03_rainfall_context.ipynb](03_rainfall_context.ipynb) (add rainfall context),
[04_under_the_hood.ipynb](04_under_the_hood.ipynb) (the lower-level building blocks).

## 1. Run the workflow

We point `run_hydroseason` at a committed case-study CSV — 21 years
(2005-2025) of DEA-derived monthly surface-water extent for the Fitzroy
River (WA) — and let it decide everything: regime, route, hydrological-year
boundaries (or events/spells), and the report.

```python
from pathlib import Path

from hydroseason import run_hydroseason

CASE_STUDY_CSV = Path("../case_studies/data/extent/fitzroy_river_wa_30m.csv")

result = run_hydroseason(
    CASE_STUDY_CSV,
    output_dir="output/01_quickstart",
    aoi_name="Fitzroy River (WA)",
)

print(f"Regime: {result.analysis.regime.regime}")
print(f"Route: {result.analysis.route}")
```

## 2. What you got back

`run_hydroseason` writes one self-contained HTML report and four CSVs
(monthly timeline, hydrological-year markers, wet events, low-extent
spells), and returns their paths on `result.artifacts`.

```python
for label, path in [
    ("HTML report", result.artifacts.html),
    ("Monthly CSV", result.artifacts.monthly_csv),
    ("Hydro-years CSV", result.artifacts.hydro_years_csv),
    ("Wet-events CSV", result.artifacts.wet_event_csv),
    ("Low-spells CSV", result.artifacts.low_spells_csv),
]:
    print(f"{label}: {Path(path).relative_to(Path.cwd())}")
```

## 3. Open the report

The HTML file is self-contained — open it directly in a browser, no server
needed. It looks like the
[live Fitzroy River example](https://tayerthiaggo.github.io/hydroseason/examples/fitzroy-river-wa.html)
linked from the README.

```python
print(f"Open this file in a browser:\n{result.artifacts.html.relative_to(Path.cwd())}")
```

## 4. Inspect the monthly CSV

Every report bundle includes a compact monthly timeline CSV — the same
`extent_pct`/`invalid_pct` series the routing decision was made from.

```python
import pandas as pd

pd.read_csv(result.artifacts.monthly_csv).head()
```
